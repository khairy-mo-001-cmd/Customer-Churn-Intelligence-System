# Business Understanding

## 1. Business Objective
The objective of this project is to build a **data-driven churn prediction and retention framework** for a UK-based cross-border online retail business operating in both **B2B and B2C segments**.

Since the business does not operate on a subscription model, customers do not formally cancel their accounts. Instead, they gradually stop purchasing over time. This means churn in this business is **implicit churn**, and therefore it must be identified through behavioral patterns rather than explicit cancellation events.

The project aims to help the business move from a **reactive retention approach** to a **proactive, segmented, and financially measurable retention strategy**.

---

## 2. Business Context
The dataset used in this project is **Online Retail II**, published through the **UCI Machine Learning Repository** by **Dr. Daqing Chen, University of Bedfordshire**.

It contains real transactional data for a **UK-registered, non-store, cross-border e-commerce business** over the period from **2009-12-01 to 2011-12-09**.

The company operates in a **hybrid wholesale and retail model**, serving:
- independent boutiques,
- gift shops,
- commercial wholesale buyers,
- and individual consumers.

The business depends heavily on:
- repeat orders,
- high transaction volume,
- stable reorder cycles,
- and seasonal purchasing behavior.

A key operational feature is the strong **Q4 seasonality**, where order volume increases significantly as wholesale customers prepare for holiday demand.

---

## 3. Current Business Problem
The business currently faces several retention-related challenges:

1. **Late churn detection**  
   Customers are often identified as lost only after a long period of inactivity, reducing the probability of successful re-engagement.

2. **Generic retention campaigns**  
   Marketing actions tend to use broad incentives rather than customer-specific or segment-specific interventions.

3. **Lack of geographic differentiation**  
   The business does not fully account for the fact that the same inactivity period may have different meanings in different markets.

4. **Limited financial linkage**  
   Retention activities are not always evaluated in terms of recovered revenue, campaign cost, and return on investment.

As a result, retention spending may be inefficient, and high-value customers may not receive the most appropriate intervention.

---

## 4. Business Goals
The project aims to:

- identify customers at risk of churn using historical transactional behavior,
- segment customers into actionable churn tiers,
- account for geographic and behavioral differences,
- support differentiated retention actions,
- and improve the financial effectiveness of retention campaigns.

---

## 5. Key Business Questions
This project is designed to answer the following business questions:

- Which customers are likely to stop purchasing in the future?
- How severe is the churn risk for each customer segment?
- Which customers represent the highest revenue at risk?
- How does churn behavior differ across the UK, Europe, and overseas markets?
- What retention action should be taken for each churn tier?
- What is the expected business return from each retention action?

---

## 6. Business Analysis

### 6.1 Business Model
The company operates as a **low-margin, high-volume** retailer with a mixed **B2B and B2C** customer base.

Profitability depends on:
- repeat wholesale orders,
- high unit volume,
- stable purchasing cycles,
- and strong seasonal demand.

This means that customer retention is not only a marketing issue, but a core revenue-protection issue.

---

### 6.2 Product Portfolio
The business offers a broad catalog of **more than 4,000 SKUs**, mainly grouped into:

- vintage and retro home decor,
- party and celebrationware,
- seasonal and holiday goods,
- stationery, crafts, and kitchenware.

This wide product range supports both retail and wholesale demand, but also means that customer behavior can vary significantly by product interest and purchase purpose.

---

### 6.3 Customer Segments
Customers can be understood in several commercially meaningful segments:

#### UK Domestic Customers
- represent the largest share of order volume,
- typically show shorter reorder cycles,
- and are more sensitive to short inactivity periods.

#### European Wholesale Customers
- usually place larger orders,
- often buy in bulk,
- and are more exposed to freight and customs considerations.

#### Overseas Bulk Buyers
- purchase less frequently,
- often order in larger shipping cycles,
- and may show long gaps between orders as part of normal business behavior.

#### B2C Customers
- usually place smaller orders,
- are more promotion-sensitive,
- and contribute less individually than wholesale clients.

#### High-Value B2B Customers
- may order less frequently but have greater commercial importance,
- and therefore require more personalized retention handling.

---

### 6.4 Geographic Behavior
The dataset includes transactions from **more than 38 countries**.

The geographic footprint can be summarized as:

- **United Kingdom:** approximately **88%–90%** of order volume  
- **Continental Europe:** approximately **7%–8%** of orders  
- **Overseas markets:** smaller share, often with bulk/container-based purchasing behavior  

This matters because churn risk should not be interpreted in the same way across all regions.

For example:
- **90 days of inactivity in the UK** may indicate meaningful churn risk.
- **90 days of inactivity in Japan or Australia** may be normal for bulk buyers with longer reorder cycles.

---

### 6.5 Seasonality
The business experiences strong seasonal effects, especially in **Q4**, when order volume rises sharply due to holiday-related wholesale stocking.

This seasonality is important because:
- short-term inactivity may sometimes reflect normal market rhythm,
- and churn interpretation must consider calendar effects.

---

### 6.6 Data Quality Considerations
The dataset contains several practical data-quality issues that are relevant to business interpretation:

- approximately **22%–24%** of transactions have missing `Customer ID`,
- invoice numbers beginning with **“C”** represent cancellations,
- and some records may reflect returns or freight-related entries.

These issues must be handled carefully to avoid distorting customer behavior analysis.

---

## 7. Competitive Landscape
The business operates in a competitive market where customers can switch suppliers with relatively low friction.

### Key competitors include:
- **Rex London**
- **Puckator**
- **Sass & Belle**
- **Amazon Business**
- **Faire & Ankorstore**

### Main competitive pressures:
- supplier switching is easy,
- wholesale buyers often work with multiple vendors,
- product freshness matters,
- freight cost and delivery reliability affect retention,
- and marketplace platforms offer strong convenience.

This means churn is not only driven by internal service quality, but also by external market alternatives.

---

## 8. Core Churn Drivers
Based on the business context, churn is likely influenced by:

1. **Low switching friction**  
   Customers can move spend to other suppliers easily.

2. **Catalog fatigue**  
   Wholesale buyers may leave if the product assortment does not refresh sufficiently.

3. **Freight and customs sensitivity**  
   Cross-border buyers may reduce orders when shipping or customs costs become less attractive.

4. **Service or availability issues**  
   Stock availability, delivery delays, or weak reorder experience may reduce repeat purchases.

5. **Competitive incentives**  
   Rivals may offer better pricing, payment terms, or logistics support.

---

## 9. Retention Strategy Logic
To make the project commercially useful, churn must be translated into **actionable retention tiers**.

The business uses fixed inactivity windows:

- **Active:** ≤ 90 days  
- **Tier 1 Early Churn:** 91–180 days  
- **Tier 2 Mid-Stage Churn:** 181–365 days  
- **Tier 3 Severe / Dormant:** > 365 days  

However, the **business response** should vary by geography and customer type.

For example:
- UK customers in Tier 1 may require fast automated re-engagement.
- European customers may respond better to freight-related incentives.
- Overseas bulk buyers may need account-manager outreach rather than aggressive discounting.

## 10. Business Value of the Project
This project is expected to create value by:

- detecting churn earlier,
- improving targeting of retention campaigns,
- reducing unnecessary discount spending,
- protecting high-value wholesale accounts,
- and linking retention activity to measurable financial outcomes.

In practical terms, the model should support better decisions about:
- **who to contact**,
- **when to contact them**,
- **what incentive to offer**,
- and **how much investment is justified**.

---

## 11. Success Criteria
The project will be considered successful if it:

- provides a clear churn segmentation framework,
- identifies at-risk customers early enough for useful intervention,
- supports market-specific retention actions,
- improves retention efficiency,
- and enables tracking of business impact.

The proposed recovery targets are:

- **Tier 1:** 10% recovery  
- **Tier 2:** 20% recovery  
- **Tier 3:** 30% recovery  

These targets provide a business benchmark for evaluating retention performance.

---

## 12. Business Constraints
Several constraints must be considered:

- missing customer identifiers limit customer-level tracking,
- cancellations and returns must be interpreted correctly,
- seasonality can affect inactivity interpretation,
- geographic differences must be incorporated into business actions,
- and retention incentives must be balanced against margin impact.

These constraints are important to ensure that the final solution is both practical and commercially realistic.

---

## 13. Business Understanding Summary
This project addresses a real business problem: identifying and retaining customers in a non-subscription, cross-border retail environment where churn is implicit rather than explicit.

The business context shows that churn is shaped by:
- customer type,
- geography,
- purchasing frequency,
- product assortment,
- competitor pressure,
- and seasonal demand.

Therefore, the goal of this phase is not only to define the problem technically, but also to establish a strong business foundation for the later data analysis and modeling stages.
