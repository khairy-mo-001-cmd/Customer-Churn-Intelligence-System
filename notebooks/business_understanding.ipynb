# 1. Executive Summary & Business Context

## 1.1 Business Model & Domain Architecture
The dataset originates from **Online Retail II**, a non-subscription-based, cross-border e-commerce enterprise operating globally. 

Unlike subscription models (e.g., SaaS or streaming services) where customer cancellation is explicitly recorded upon subscription termination, non-subscription e-commerce experiences **Implicit Churn**. Customers cease purchasing without issuing a formal notification, making churn identification a dynamic probabilistic problem rather than a discrete logging event.

## 1.2 Strategic Business Objective
The primary business objective is to transition from reactive customer acquisition to proactive retention engineering. 

By predicting churn risk across temporal tiers, the marketing and customer success teams can deploy targeted, tiered retention interventions. This minimizes churn mitigation costs while maximizing **Customer Lifetime Value (CLV)** and protecting **Revenue at Risk**.

# 2. Multi-Tier Churn Framework & Actionable Strategy

Rather than treating churn as a binary state, this system implements a **Graduated Multi-Tier Churn Taxonomy**. The severity of inactivity dictates the business intervention type and allocated financial discount budget:

| Tier Level | Inactivity Window | Churn Severity | Risk Classification | Retention Intervention Strategy | Target Win-Back Rate | Discount Budget |
| :--- | :--- | :--- | :--- | :--- | :--- | :--- |
| **Active** | $\le 90$ Days | Operational | Low Risk / Engaged | Standard promotional newsletters & loyalty rewards | N/A (Baseline) | $0\%$ |
| **Tier 1** | $91 - 180$ Days | Early Stage | Moderate Risk | Automated email reminders & low-friction incentives | $10\%$ Recovery | $5\% - 10\%$ Coupon |
| **Tier 2** | $181 - 365$ Days | Mid Stage | High Risk | Dynamic personalized recommendations & targeted surveys | $20\%$ Recovery | $15\% - 20\%$ Discount |
| **Tier 3** | $> 365$ Days | Severe / Dormant | Lost Customer | Aggressive win-back campaigns, VIP outreach, or survey removal | $30\%$ Recovery | $25\% - 30\%$ Special Offer |
